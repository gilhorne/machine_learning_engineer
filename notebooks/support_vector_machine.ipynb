# Support Vector Machines (SVM)

- powerful supervised machine learning model used for classification.

- Makes classifications by defining a decision boundary, then seeing what side of the boundary an unclassified point fall on.

- decision boundaries are defined by using a training set of classified points. 

- *Seperating Hyperplane* :more than 3 features.

## Optimal Decision Boundaries

![decision boundaries](images/decision_boundaries.png)

- Decision boundaries should be as far away from training points as possible.
    - Maximizing the distance between the decision boundary and points in each class will decrease the choice of false classification.



## Support Vectors and Margins

- `Support Vectors` :points in the training set closest to the decision boundary. (*these vectors are what define the decision boundary*)

- Instead of thinking of the training data as points, think of them as vectors coming from an origin.

![support vectors](images/support_vectors.png)

*if you are using `n` features, there are at least `n+1` support vectors*

- the distance between a support vector and the decision boundary is called the *margin* 
    - The margin should be as large as possible

![margins](images/margins.png)

- SVMs are fast because they only use support vectos and ignore the irrelevant training points.

## SVM - Scikit-learn (Support Vector Classifier - SVC)

```python
# import library (scikit-learn) 
from sklearn.svm import SVC
from graph import points, labels

# 1. Create SVC object:
classifier = SVC(kernel = 'linear')

# 2. Train model on list of data points, 
# and list of labels associated with those points.
    # - labels are analogous to the color of the point
    # - 1 is equal to a red point 
    # - 0 is equal to a blue point
# .fit()

training_points = [[1, 2], [1, 5], [2, 2], [7, 5], [9, 4], [8, 2]]
labels = [1, 1, 1, 0, 0, 0]
classifier.fit(training_points, labels)
```
*- calling `.fit()` creates the line between the points*

![output example](images/output_example_1.png)

```python

# 3. Use .predict() to predict label of new points. 
    # - .predict() method takes a list of points you want to classify

print(classifier.predict([[3,2]]))
```

![output example](images/output_example_2.png)

- Print `classifier.support_vectors_` :to see which points from the training set are the support vectors. 

## Outliers

- The size of the margin decreases when a sinle outlier is presented.
- As as result, the decision boundary changes as well.

*image*
![outlier Margin example](images/outlier_margin_example.png)

SVMs have a parameter `C` that determines how much error the SVM will allow.

- If `C` is large, the SVM has a hard margin
    - It won't allow for misclassification
    - Results in a fairly smaller margin

- If `C` is too large, the model runs the risk of overfitting
    - SVM rely too heavily on the training data, including outliers

- If `C` is small, the SVM has a soft margin
    - Some points might fall on the wrong side of the line,
    - But the margin will be large, making it more resistant to outliers

- If `C` is too small, the model runs the riskc of underfitting
    - SVM will allow so much error that the training data won't be represented

```python
# Create classifier object with SVC
# Set the value C

classifier = SVC(C = 0.01)
```

**Note**

- The optimal value of `C` will depend on your data. 
- Don’t always maximize margin size at the expense of error. 
- Don’t always minimize error at the expense of margin size. 
- The best strategy is to validate your model by testing many different values for `C`.

## Polynomial Kernel: (Code)

```python
from sklearn.datasets import make_circles
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split

# Makes concentric circles
points, labels = make_circles(n_samples=300, factor=.2, noise=.05, random_state = 1)

# Makes training set and validation set.
training_data, validation_data, training_labels, validation_labels = train_test_split(points, labels, train_size = 0.8, test_size = 0.2, random_state = 100)

classifier = SVC(kernel = "linear", random_state = 1)
classifier.fit(training_data, training_labels)
print(classifier.score(validation_data, validation_labels))

# print(training_data[0])
# create two empty lists:
new_training = []
new_validation = []

# loop through every point in training_data,
# append a list to new_ training
# The square root of 2 times point[0] times point[1].
# point[0] squared.
# point[1] squared.
for point in training_data:
  new_training.append([2**0.5*point[0]*point[1], point[0]**2, point[1]**2])

# loop through validation_data:
for point in validation_data:
  new_validation.append([2**0.5*point[0]*point[1], point[0]**2, point[1]**2])

# Retrain classifier model:
classifier.fit(new_training, training_labels)
print(classifier.score(new_validation, validation_labels))
```


## Radial Basis Function Kernel (RBF)
- Default kernel used in scikit-learn's SVC object
- Transforms two-dimensional points into points with an infinite number of dimensions.

- `gamma` parameter: Can be tuned to be more or less sensitive to the training data.
    - higher gamma will put more importance on training data and could result in overfitting.

    - lower gamma makes points in training data less relevant and can result in underfitting.

```python
classifier = SVC(kernel = "rbf", gamma = 0.5, C = 2)
```
