```python
# Inheritance: 
# Usage of methods from parent classes:

# PARENT CLASS:
class Parent():
    # Create trait var: (List of traits for randomzation)
    trait1 = 'talkative'
    trait2 = 'hyper'
    trait3 = 'shyness'
    trait4 = 'coolness'

    def __init__(self):
        pass
    
    def trait(self):
        self.trait = Parent.trait1 #change trait number for different trait output.
        # print('A {} trait was passed down.'.format(self.trait))
        print(f'A {self.trait} trait was passed down.')

# CHILD CLASS:
# The Child class inherited the .trait() class variable from the Parent class.
class Child(Parent):
    # Overriding methods: Update the trait method based on child class.
    pass


# OUTPUT:
child1 = Child()
child1.trait()
```

```python
# Using the .super() method for subclass
# Override the subclass class var with an additional trait.
# GRANDCHILD CLASS:
class GrandChild(Parent):
    def trait(self):
        super().trait()
        self.trait = Parent.trait3
        print(f'The {self.trait} trait was also passed down.')

gchild1 = GrandChild()
gchild1.trait()
```

```python
class CoolGChild(GrandChild):
    print(f'The {Parent.trait4} trait is the main')

gchild2 = CoolGChild()
gchild2.trait()
```

```python
# Polymorphism:
# operator overloading 
# dunder methods 

class Employee():
  new_id = 1
  def __init__(self):
    self.id = Employee.new_id
    Employee.new_id += 1

class Meeting:
  def __init__(self):
    self.attendees = []
  
  def __add__(self, employee):
    print("ID {} added.".format(employee.id))
    self.attendees.append(employee)

  # create len() dunder method to count len of self.attendees. 
  def __len__(self):
    return len(self.attendees)
  
    
e1 = Employee()
e2 = Employee()
e3 = Employee()
m1 = Meeting()

# add the employees to the meeting. 
m1 + e1
m1 + e2
m1 + e3

print(len(m1))
```

```python
# Abstraction:
# Abstract class can not be instantiated. 

from abc import ABC, abstractmethod

class AbstractEmployee(ABC):
  new_id = 1
  def __init__(self):
    self.id = AbstractEmployee.new_id
    AbstractEmployee.new_id += 1

# Abstract method
  @abstractmethod
  def say_id(self):
    pass

# Add abstract method.
class Employee(AbstractEmployee):
    def say_id(self):
      return 'My Id number is: {}'.format(self.id)

e1 = Employee()
print(e1.say_id())
```

```python
# Encapsulation:
# the process of making methods and data hidden inside the object.

class Employee():
    def __init__(self):
        self.id = None #public 
        # self.id encapsulation:
        self._id = None #protected, this notates it is intended to be used only within the module.
        self.__id = None #private

e = Employee()
print(dir(e))
```

```python
# Encapsulation:
# Getter, Setter, Deleter Functions

# These functions are useful in making sure that the data being handled is appropriate for the defined class functionality.

class Animal:
  def __init__(self, name):
    self._name = name
    self._age = None

  def get_age(self):
    return self._age

  def set_age(self, new_age):
    if isinstance(new_age, int):
      self._age = new_age
    else:
      raise TypeError

  def delete_age(self):
    del self._age
    print("_age Deleted")


# OUTPUT:
a = Animal("Rufus")
print(a.get_age()) # None

a.set_age(10)
print(a.get_age()) # 10

a.set_age("Ten") # Raises a TypeError

a.delete_age() # "_age Deleted"
print(a.get_age()) # Raises a AttributeError
```
