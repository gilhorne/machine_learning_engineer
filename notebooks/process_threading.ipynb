# Joining a thread

`.join()` : tells one thread to wait for this thread to stop before moving on.
 - used after each thread has been initiated.

> Method 1
 ```python
import threading

t1 = threading.Thread(target=function1, args=(arg1,))
t2 = threading.Thread(targer=function2, args=(arg2,))
t3 = threading.Thread(target=function3, args=(arg3,))

#initiate
t1.start()
t2.start()
t3.start()

#join
t1.join()
t2.join()
t3.join()
```


> Method 2

**:: more efficient approach**

```python

import threading

threads = []

args = [arg1, arg2, arg3]

for i in range(len(args)):
    t = threading.Thread(target=function, args=(args[i]))
    threads.append(t)
    #initiate
    t.start()

#join
for t in threads:
    t.join()
```
