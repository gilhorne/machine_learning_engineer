# Naive Bayes Classifier: Classifier

***USING `MultinomialNB` CLASSIFIER***

*first*
- Can be trained by using `.fit(array_1, array_2)` method that takes **two** parameters.
    - the array of data points
    - the array of labels corresponding to each data point

*once model has been trained*
- `.predict()` :takes a list of points to classify; returns the predicted labels of those points.

*to return probability of each label given a point*
- `.predict_proba()` :will return the probability of each label given a point.
    - will return the liklihood of a good or bad review, not just if it was good or bad.

```python
from reviews import counter, training_counts
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

review = "This crib is great"
review_counts = counter.transform([review])

#create MultinomialNB object
classifier = MultinomialNB()

#create a list of training points, 1000 [0]s (for negative) and 1000 [1]s (for positive)
training_labels = [0] * 1000 + [1] * 1000

#.fit() the classifier
classifier.fit(training_counts, training_labels)

#print and call .predict()
#print(classifier.predict(review_counts))

#print and call .predict_proba()
print(classifier.predict_proba(review_counts))

#print text output
if classifier.predict(review_counts) == 0:
  print('Review is negative.')
else:
  print('Review is positive.')
```
