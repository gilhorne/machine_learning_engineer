# Bayes Theorem

![bayes theorem](images/bayes_theorem.png)


- `P(B|A)` :The probability that event `B` will happen given that event `A` has already happened.

---
- **Independent Events** :if two events are independent, then the occurrence of one event does not affect the probabiity of the other event.

![independent events](images/independent_events.png)

- **Dependent Events** :if two events are dependent, then when one event occurs, the probability of the other event occuring changes in a predictable way.

![dependent events](images/dependent_events.png)

### Conditional Probability

- The probablity that two events happened.
*easiest to calculate when two events are independent*

![conditional probability](images/condtional_probability.png)

*example*

```python
# This week, there is a 30% probability that it will rain on any given day. At a certain high school, gym class is held on three days out of the five day school week.

# On a school day, what is the probability that it is raining and the students have gym class?
```

```python
p_rain = 0.30
p_gym = 3.0 / 5.0

p_rain_and_gym = p_rain * p_gym

print(p_rain_and_gym)
```

- Smoothing

![image_4](images/image_4.png)

```python

from reviews import neg_counter, pos_counter

review = "This cribb was amazing"

percent_pos = 0.5
percent_neg = 0.5

total_pos = sum(pos_counter.values())
total_neg = sum(neg_counter.values())

pos_probability = 1
neg_probability = 1

review_words = review.split()

for word in review_words:
  word_in_pos = pos_counter[word]
  word_in_neg = neg_counter[word]
  
  #Smooting:
  pos_probability *= (word_in_pos + 1) / (total_pos + len(pos_counter)) 
  neg_probability *= (word_in_neg + 1) / (total_neg + len(neg_counter))
  
print(pos_probability)
print(neg_probability)
```
