# Asyncio Module

> *keywords that allow building and executing asynchronous code in the program*

- `async` : declares a function as a *coroutine*
    - *coroutines* are functions that may return normally with a value,
    or may suspend themselves internally and return a continuation.

- `await` : 
    - suspends execution of the current task until whatever is being awaited
    on is completed.

> `hello_async()` program

```python
import asyncio

#coroutine
async def hello_async():
    print('hello')
    await asyncio.sleep(3) #waits 3 seconds
    print('How are you?')

#run coroutine
loop = asyncio.get_event_loop()
loop.run_until_complete(hello_async())

# Python 3.7 syntax update
# asyncio.run(hello_async)
```

## Multiple Asynchronous Tasks

```python
async def main(): #1
    tasks = [task1(arg1), task2(arg2), task3(arg3)] #2
    await asyncio.gather(*tasks) #3-4
```

1. define `main()` as coroutine function.

2. `tasks` is a list of seperate fuction calls.
> *note: each of `task1(), task2(), task3()` are coroutine functions.*

3. `asyncio.gather()` groups all of the tasks together to run concurrently,
given `await` syntax.

4. `*tasks` unpacks the task lists.
