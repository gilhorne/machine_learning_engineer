- Function must be deterministic; for any given input, the function must return the same output. 

- Recursion is used to execute code repeatedly.

```python
# Functional Programming: Reusable programming

lst = [1,2,3,4,5]

# define function:
def app(l):
    ret = list(l)
    ret[0] = 0
    ret[-1] = 0
    return ret

# Function can be called

zero_list = app(lst)
print(zero_list)
```

```python
# Recursion:

# def recursive_function(n):
    # Base case to terminate recursion goes here
    # Call to recursive_function with new parameter goes here

# TEST:
# recursive function in the computation of the nth Fibonacci number done like so:

def fib(n):
  if n <= 1:
    return 1
  return fib(n-1) + fib(n-2)
```

- Functions can be passed as arguments within other functions. 

- This treats functions as first-class citizens. 

In programming, “first-class citizens” means that something (like functions, objects, or variables) can be used just like any other value. For example, if functions are first-class citizens, you can:

- Assign them to variables
- Pass them as arguments to other functions
- Return them from functions
- Store them in data structures

This gives you more flexibility when writing code. In Python, functions are first-class citizens.

```python
def add(x, y): # function
  return x + y

def sub(x, y): # function
  return x - y

def times3(a, b, function): #function passing another function as argument
  return 3 * function(a, b)

#RESULTS:
add_then_times3 = times3(2, 4, add) # 18
sub_then_times3 = times3(2, 4, sub) # -6
```

```python
# lamda functions:

# Lambda expressions are commonly used for defining single-use functions which are then passed to higher-order functions.
# lambda parameter_list : expression

def times3(a, b, function):
  return 3 * function(a, b)

add_then_times3 = times3(2, 4, lambda x, y: x + y) # 18
sub_then_times3 = times3(2, 4, lambda x, y: x - y) # -6
```

```python
# Immutable datatypes

from collections import namedtuple #immutable tuple for reusabability.

# Create namedtuple() called country.
country = namedtuple('country', ['name','capital', 'continent'])

# Create tuple instances.
france = country('France', 'Paris', 'Europe')
japan = country('Japan', 'Tokyo', 'Asia')
senegal = country('Senegal', 'Dakar', 'Africa')

# Create tuple of country names, print result.
countries = (france.name, japan.name, senegal.name)
# capitals = [i.capital for i in countries]
print(f'Countries are {countries[0]}, {countries[1]} and {countries[2]}.')
```
