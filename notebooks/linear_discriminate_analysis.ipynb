# Linear Discriminate Analysis:

- Linear Discriminate Analysis (LDA): Dimensionality reduction technnique for classification problems with continuous independent variables.

    - can be used for:
        - image recognition
        - marketing
        - biological classifiaction

- finds the best possible subspace for a given classification probelem.

---

## LDA With scikit-learn :

- LDA 1-dimensional subspace:
    `LinearDiscriminantAnalysis(n_components=1)`

*Note: scikit-learn requires that `n_components` is less than the number of classes minus one and less than the number of features.*

```python
# Import libraries:
import pandas as pd
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# load data
seed = pd.read_csv('seeds.csv')
X = seeds.drop('variety' , axis=1)
y = seeds['variety']

# create LDA model
lda = LinearDiscriminantAnalysis(n_components=1)

# fit the data and create a subspace X_new
X_new = lda.fit_transform(X,y)
```

---

`X_new` is a 1-dimensional subspace of X. 

- Can train a classifier :

```python 
# Import library
from sklearn.linear_model import LogisticRegression

# Create logistic regression model
lr = LogisticRegression()

# Fit the model
lr.fit(X_new, y)

# Model accuracy
print(lr.score(X_new, y))
```

## Assumptions:

- LDA works if the size of the smallest class is greater than the number of variables.

- LDA relies on the assuption that independent variables in each class are multivariate normal.

- Homoscedasticity : is the assumption that each class has the same covariance matrix.

    - Quadratic Discriminate Analysis (QDA): can be used when classes have different covariance matrices.

- Independent veriables are not mutlicolinear.
    - no variable is closely correlated with any other variable or linear combination of variables. 

- Observation in data set should be statistically independent. 

- LDA can be sensitive to outliers
    - best practice is to remove extreme outliers before doing LDA.

- LDA is prone to overfitting:
    - `shrinkage` parameter is a regularization technique used to help prevent overfitting.
