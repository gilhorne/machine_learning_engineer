Logging:

- By logging messages, we can easily debug our software and investigate events during the development, testing, and production phases.

```python
# Logging:
#  1: Import logging module and sys library into application.

import logging
import sys
```

```python
# 2: Obtain logger object using getLogger(name) method. 
# - This method takes in an optional, single input parameter called name.
# - This represents the name of the logger.
# - Calling getLogger() with the same name value returns the same logger object.
# - If there is no name, the root logger is returned.

# Create var "logger", set to logging.getLogger(__name__)

logger = logging.getLogger(__name__)
```

```python
# 3: Inform the logger where we want the logging message to go.
# - Use logging module's StreamHandler class. (Handles where our logged messages output)
# - StreamHandler takes in optional, single output, if value is not supplied, logs direct to sys.stderr
# - To output directly to console, supply sys.stdout
# - Make sure to import sys library

stream_handler = logging.StreamHandler(sys.stdout)
```

```python
# 4: Add stream handler object to the logger.
# - Use addHandler(hdlr) method to add a specific handler to logger object
# - hdlr represents the handler object to add.

logger.addHandler(stream_handler)
```

Log Levels:

- Each logging level is a constant within the logging module with an associated numeric value. 
- The higher this numeric value, the higher the severity of the log message.

Levels:
- logging.NOTSET (0): 
    - searches for the first non-NOTSET ancestor logger and inherits its logging level. 
    - If there is none, our output threshold will be zero.

- logging.DEBUG (10): 
    - provide detailed information that is useful for debugging the application. 
    - Numeric value of 10.

- logging.INFO (20): 
    - general operations where expected information or output is logged. 
    - Numeric value of 20.

- logging.WARNING (30): 
    - alert to a current or impending, unexpected issue or error. 
    - This logging level does mean that the software or application will continue to run despite the warning message. 
    - Numeric value of 30.

- logging.ERROR (40): 
    - indicate serious problems that cause functionality within the software or application to break. 
    - Numeric value of 40.

- logging.CRITICAL (50): most severe of errors and issues. 
    - These errors indicate that the software or application may stop running altogether. 
    - Numeric value of 50.

```python
print(logging.DEBUG) #print logger value = (10)
```

The logging module has several methods that we can use to log messages and errors with an assigned severity level. Those methods are:

- debug(msg) which logs a message with level DEBUG

- info(msg) which logs a message with level INFO

- warning(msg) which logs a message with level WARNING

- error(msg) which logs a message with level ERROR

- critical(msg) which logs a message with level CRITICAL

```python
# Using our logger object, we can call these methods directly while inputting our desired log message as the parameter.

logger.debug('This is a debug messgage!')

# The logging module also provides a method log(level, msg) that allows us to log a specific log level and message. 
# The level parameter represents the numeric log level to use, and the msg parameter is the string value that will be logged.

logger.log(logging.INFO, "This is an info message!")
```

Setting the Log Level:

When a log level is set for a logger, any log message that has this log level and higher will output.

To change the default log level for a logger, we can use the method setLevel(level)

- The level parameter represents the numeric value of the log level to use.

```python
logger.setLevel(logging.DEBUG)
```

USE CASES:

- Troubleshooting a problem within our application that is causing it to crash, it would be optimal to see only the CRITICAL level log messages and filter out all other messages.

- Debugging often, setting the log level to DEBUG is useful for obtaining helpful debugging log messages.

```python
# To write logs to a saved file, we can use the logging module class FileHandler(filename). 

# The filename initialization parameter is a string value that represents the filename for the written log file.

# we can add the FileHandler object to the logger by using the addHandler(hdlr) method.

file_handler = logging.FileHandler("my_program.log")
logger.addHandler(file_handler)
```

```python
#  Adding multiple addHandler(hdlr) calls for each handler object.

logger.addHandler(file_handler) # file log
logger.addHandler(stream_handler) # console log
```

Formatting Logs:

This allows us to include helpful information with each logged message, including:

If no custom formatting is specified, Python uses the default formatting for all log messages:

%(levelname)s:%(name)s:%(message)s

- %(levelname)s is a string that represents the log level name

- %(name)s represents the string of the name of the logger

- %(message)s is the logged message string.

```python
# We can create a custom formatter object using the logging module’s Formatter class, which accepts the formatted string as the first input value:

formatter = logging.Formatter("[%(asctime)s] %(levelname)s:%(name)s:%(lineno)d:%(message)s")

# set this formatter object to logger, use the setFormatter(fmt) method where the fmt parameter is the Formatter object

stream_handler.setFormatter(formatter)
```

Using basicConfig():

The basicConfig() method allows for the basic configuration of the logger object by configuring the log level, any handlers, log message formatting options, and more in one line of code. 

```python
logging.basicConfig(filename="calculator.log", level=logging.DEBUG, format="[%(asctime)s] %(levelname)s - %(message)s")

# This code automatically adds a FileHandler object that will write to the filename calculator.log, set the log level to DEBUG, and format all logged messages to have a timestamp, log level, and message information.
```
