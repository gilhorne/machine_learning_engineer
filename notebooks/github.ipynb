# Git Branch

`git branch` :returns the current branch

`git branch new_branch` :creates new branch

*branch name should describe the purpose*

`git checkout branch_name` :switches to branch

`git add filename` :adds files to staging area

`git commit -m "Commit Message"` :commit with message

`git merge branch_name` :merges branch to master

`git branch -d branch_name` :deletes branch after edits have been made

### Git Remote

![workflow](images/workflow.png)

- A shared Git repository that allows mulitple collaborators to work on the same Git project from different locations.

`git clone` :get replica of repository

`git clone remote_location clone_name`

- `remote_location` :tells Git where to go to find the remote.
*could be web address or filepath*

- `clone_name` :the name you give to the directory in which Git will clone the repository.

- `git remote -v` :list of Git projects remotes.

- `git fetch` :see if changes have been made, and bring changes to local copy.

- `git merge origin/master` :integrates origin/master into local master branch.

- `git push origin branch_name` :push branch to the remote, origin, where it can be reviewed and merged into the master branch.
