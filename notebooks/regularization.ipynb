# Regularization:

- Statistical technique that minimizes overfitting
- Executed during the model fitting step
- Embedded feature selection method due to being implemented while parameters of model are being calculated

*Question: How well can the model generalize from known to unknown data?*

- Regularization makes sure that model is still accurate

- bias-variance tradeoff
- hyperparameter tuning

- *What is underfitting?*
    - least representative of the data, touching least data points

- *what is overfitting?*
    - uses the most datapoints, sensitive to small variations
    - If a model is able to represent a particular set of data points effectively but is not able to fit new data well, it is overfitting the data.

        - It fits the training data well but performs significantly worse on test data

        - It typically has more parameters than necessary, i.e., it has high model complexity

        - It might be fitting for features that are multi-collinear (i.e., features that are highly negatively or positively correlated)

        - It might be fitting the noise in the data and likely mistaking the noise for features

### Loss Function 

![loss function](images/loss_function.png)

#### Pythonic Code: Loss Function:

```python
# Loss Function
def loss_function(b1,b2,y,x1,x2):
    error = y - b1*x1 - b2*x2
    loss = np.mean(error**2)
    return loss
```

![loss function graph](images/loss_function_graph.png)

- Each ellipse in this plot represents the points where the loss function has the same value. The larger the ellipse, the higher the value of loss for all the points (b1,b2) that fall on the ellipse, as we can see from the labels on the contours.

- Following this logic, the center of this set of concentric ellipses should represent the minimum value of the loss function and the value of the coefficients (b1,b2) at this point should be the solution of the regression problem and represent the best fit for our data.

## Regularization Terminology:

- Regularization penalizes models for overfitting by adding a "penalty term" to the loss function.

     Regularization Terms ("penalty terms"):

    ![regularizaion terms](images/regularization_summary.png)

    - Lasso (L1) regularization:
        - ![L1 Regularization](images/L1_regularization.png)

    - Ridge (L2) regularization:
        - ![L2 Regularization](images/L2_regularization.png)

    - *multiplied by a factor `alpha` and added to the old loss function*

    new_loss_function = old_loss_function + a * regularization_term 

    - sum of the magnitudes of a vector -> L1 norm -> L1 regularization
    - square root of the sum of the magnitudes -> L2 norm -> L2 regularization

**Lasso** : **“Least Absolute Shrinkage and Selection Operator”** 

- it shrinks the absolute value of coefficients and selects parameters by way of setting some of the coefficients to zero. 

- If the coefficient of a regression model is set to zero, it means that we’ve pretty much eliminated the feature associated with it. 

- In models with a high number of features, lasso regularization tends to set a significant fraction of its parameters to zero, thus acting as a feature selection method.

#### Code Example:
```python
import pandas as pd
import numpy as np
import codecademylib3
import matplotlib.pyplot as plt
import helpers

df = pd.read_csv("./student_math.csv")
y = df['Final_Grade']
X = df.drop(columns = ['Final_Grade'])

# 1. Train-test split and fitting an l1-regularized regression model
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
lasso = Lasso(alpha = 0.1)
lasso.fit(X_train, y_train)

l1_pred_train = lasso.predict(X_train)
l1_mse_train = np.mean((l1_pred_train - y_train)**2)
print("Lasso (L1) Training Error: ", l1_mse_train)

# 2. Calculate testing error
l1_pred_test = lasso.predict(X_test)
l1_mse_test = np.mean((l1_pred_test - y_test)**2)

print(l1_mse_test)

# 3. Plotting the Coefficients
predictors = X.columns
coef = pd.Series(lasso.coef_,predictors).sort_values()
plt.figure(figsize = (12,8))
plt.ylim(-1.0,1.0)
coef.plot(kind='bar', title='Regression Coefficients with Lasso (L1) Regularization')

plt.show()

```

### L2 - Ridge Regulariztion:

- Mathematical formulation of the "Ridge" regularization belongs to a class of analysis methods known as "ridge analysis'

![L2 Ridge Regularization](images/l2_regularization_function.png)

![L2 Regularization](images/l2_regularization_graph.png)

![L2 Summary](images/l2_summary.png)

- Ridge coefficients can never be exactly zero. 

- L2 regularization is therefore not a feature elimination method like L1. 

- The coefficients of L2 get arbitrarily small but never zero. 

*This is particularly useful when we don’t want to get rid of features during modeling but nonetheless want their relative importances emphasized.*

![L2 Equation](images/l2_regularization_equation.png)


```python
import pandas as pd
import numpy as np
import codecademylib3
import matplotlib.pyplot as plt
import helpers

df = pd.read_csv("./student_math.csv")
y = df['Final_Grade']
X = df.drop(columns = ['Final_Grade'])

# 1. Train-test split and fitting an l2-regularized regression model
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
ridge = Ridge(alpha = 100)
ridge.fit(X_train, y_train)

#Training error
l2_pred_train = ridge.predict(X_train)
l2_mse_train = np.mean((l2_pred_train - y_train)**2)
print("Ridge (L2) Training Error: ", l2_mse_train)

# 2. Calculate testing error
l2_pred_test = ridge.predict(X_test)
l2_mse_test = np.mean((l2_pred_test - y_test)**2)
print("Ridge (L2) Testing Error: ", l2_mse_test)

# 3. Plotting the Coefficients
predictors = X.columns
coef = pd.Series(ridge.coef_,predictors).sort_values()
plt.figure(figsize = (12,8))
plt.ylim(-1.0,1.0)
coef.plot(kind='bar', title='Regression Coefficients with Lasso (L1) Regularization')

plt.show()
```

### Hyperparameter Tuning: alpha

`alpha` : known as a hyperparameter in machine learning.

- chosen prior to fitting the model and is used to control the learning process. 

- determines how much we want to control for overfitting during the model fitting step in regularization. 

- The `alpha` ia inversely proportional to `s`, in the case of L1 regularization.
    - `s^2`, in the case of L2 regularization

*The larger the alpha, the smaller the size of the allowed coefficients*

- If `s` is very large (i.e., `alpha` is very small), the unregularized loss function minimum can easily fall within this large boundary and thus make it similar to regression without regularization.

- If `s` is very small (i.e., `alpha` is very large), the regression coefficients become very small and the loss value for the best fit becomes large making the regression over-regularized.

*hyperparameter tuning : the process to find the right amount of regularization by tuning `alpha`.*


#### Bias-Vaiance Tradeoff:

- Regularization term within the loss function introduces bias to our problem

    - biasing the model places coeficients within the regulariation boundary
    - the greater the `alpha`, the smaller the coefficients and the more biased the model.

- perform regularization to prevent the model from overfitting the data. 

- If a model have high variance, it is likely fitting for random errors or noise within the data. 

- introduce the bias term to minimize the variance. 

*allowing it to get arbitrarily large will run the risk of underfitting the model*

**NOTE**

- the machine learning model should have low bias and low variance :
    - model should perform well on training data as well as test data

**bias-variance tradeoff** :
- lowing either bias or variance, raises the other.

- hyperparameter tuning helps find a spot in this tradeoff to ensure that neither bias nor variance gets too high.

    - hyperparameter tuning is typically performed on the validation set.
