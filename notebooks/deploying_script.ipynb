# Deploy Programs with Flask (Python)

- What is deployment?

    *A production state process after designing, building and thoroughly testing the application.*

# Flask
- A lightweight framework for developing web applications. 
- Provides an efficient way to deploy web-based projects.
- Offers a built_in development server that an be used to test and deploy web application code locally.

### features:
- templating : renders HTML templates and provide functional back-end code for template components

---

**CODE**

```python
# import Flask library 

app = Flask(__name__) #1

@app.route('/') #2-3
def welcome():
    return "Welcome!"
```

1. Create Flask instance and assign it to the variable named `app`

2. `@app.route` decorator creates several routes
    - routes correspond to the URL web request made to access different pages of the web application. Main URL of the application is denoted by `/`
    - to add additional URL routes `/route_name`

```python
app.run()
```

3. `app.run()` starts the application server, allowing the application to load in the web browser.

*flask --app app run* to run in terminal

- program will deploy on localhost
*localhost/route* will route to function pages
